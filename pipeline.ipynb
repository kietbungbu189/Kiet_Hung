```python
#1.EXTRACT DIM TABLES (PRODUCT, SUPPLIER, LOGISTICS) FROM TRANSACTION TABLE
#import panda
import pandas as pd
# Save this file as peek_csv.py in your dataset folder
import csv

#1.1. import from this location of the exact name of the file to extract:
target_fileoraddress = "C:/Users/USER/Downloads/study docs/Comp604/assignment/assignment2/New Version/fact_supply-chain-data.csv"

print(f"Printing the first 3 rows of {target_fileoraddress}")

#1.2. Sending request to internet or opening local file
# Use 'errors=ignore' to bypass any messy formatting or bad characters during read

#this is open using python without panda:
#with open(target_fileoraddress, mode = 'r') as response_or_file:
    #line_or_data = response_or_file.readline()
#this is open file with panda:
response_or_file = pd.read_csv(target_fileoraddress)

#1.3. open destination:
target_fileoraddress2 = "C:/Users/USER/Downloads/study docs/Comp604/assignment/assignment2/New Version/dim_Product.csv"
target_fileoraddress3 = "C:/Users/USER/Downloads/study docs/Comp604/assignment/assignment2/New Version/dim_Supplier.csv"
target_fileoraddress4 = "C:/Users/USER/Downloads/study docs/Comp604/assignment/assignment2/New Version/dim_Logistics.csv"

#1.3.1 clean duplication in response_or_file:
response_or_file3 = response_or_file.drop_duplicates(['Supplier name', 'Location'])
response_or_file4 = response_or_file.drop_duplicates(['Shipping carriers', 'Transportation modes', 'Routes'])

#1.4. store & assign retrieved data to a variable: (add extra square brackets for data from internet, eliminate for local file)
flatten_data2 = {'SKU': response_or_file['SKU'], 'Product type': response_or_file['Product type'], 'Price': response_or_file['Price']}
unique_suppliers3 = len(response_or_file3)
flatten_data3 = {'Supplier index': range(unique_suppliers3), 'Supplier name': response_or_file3['Supplier name'], 'Location': response_or_file3['Location']}
unique_logistics4 = len(response_or_file4)
flatten_data4 = {'Logistics index': range(unique_logistics4), 'Shipping carriers': response_or_file4['Shipping carriers'], 'Transportation modes': response_or_file4['Transportation modes'], 'Routes': response_or_file4['Routes']}

#1.5 turn flatten data in variable into panda table:
table2 = pd.DataFrame(flatten_data2)
table3 = pd.DataFrame(flatten_data3)
table4 = pd.DataFrame(flatten_data4)

#1.6. write panda-table-data from varialbe to destination file:
table2.to_csv(target_fileoraddress2, index = False) #deleted " mode = 'a', " because append will cause script to add new data to old each run time
table3.to_csv(target_fileoraddress3, index = False) #deleted " mode = 'a', " because append will cause script to add new data to old each run time
table4.to_csv(target_fileoraddress4, index = False) #deleted " mode = 'a', " because append will cause script to add new data to old each run time


#2.MERGE TO CREATE SUPPLIER INDEX AND LOGISTICS INDEX COLUMNS IN THE TRANSACTION TABLE
#2.1 load 2 file dimension and fact:
target_fileoraddress5 = "C:/Users/USER/Downloads/study docs/Comp604/assignment/assignment2/New Version/fact_supply-chain-data.csv"
target_fileoraddress6 = "C:/Users/USER/Downloads/study docs/Comp604/assignment/assignment2/New Version/dim_Supplier.csv"
target_fileoraddress7 = "C:/Users/USER/Downloads/study docs/Comp604/assignment/assignment2/New Version/dim_Logistics.csv"

#2.2 open target:
with open(target_fileoraddress5, mode='r') as response_or_file5, \
    open(target_fileoraddress6, mode='r') as response_or_file6, \
    open(target_fileoraddress7, mode='r') as response_or_file7:
    line_or_data5 = pd.read_csv(response_or_file5) 
    line_or_data6 = pd.read_csv(response_or_file6)
    line_or_data7 = pd.read_csv(response_or_file7)

#2.4 merge the 2 tables and save & assign retrieved data to a variable:
#line_or_data1 comes first, and become 'left' table, line_or_data2 hence 'right' --> use left merge to protect the fact table
merge6 = pd.merge(line_or_data5, line_or_data6, on=['Supplier name', 'Location'], how='left')
merge7 = pd.merge(merge6, line_or_data7, on=['Shipping carriers', 'Transportation modes', 'Routes'], how='left')

#2.4.1 drop the old colums before writing to new file:
merge7 = merge7.drop(columns = ['Supplier name', 'Location', 'Shipping carriers', 'Transportation modes', 'Routes'])


#2.6 write table-data into destination file:
target_fileoraddress8 = "C:/Users/USER/Downloads/study docs/Comp604/assignment/assignment2/New Version/fact_supply-chain-data_updated3.csv"
merge7.to_csv(target_fileoraddress8, index = False)
```
